```python
from vpython import *
```

# Visual parameters

```python
VECTOR_O = vector(0, 0, 0)
VECTOR_U_X = vector(1, 0, 0)
VECTOR_U_Y = vector(0, 1, 0)
VECTOR_U_Z = vector(0, 0, 1)

BALL_RADIUS = 2
BALL_COLOR = color.yellow
BALL_OPACITY = 1
BALL_AXIS = VECTOR_U_X
BALL_UP = VECTOR_U_Y

DIR_VECTOR_LENGTH = 5
DIR_VECTOR_X_COLOR = color.red
DIR_VECTOR_Y_COLOR = color.green
DIR_VECTOR_Z_COLOR = color.blue

SIZE_UNIT = 8
CAMERA_AXIS_DEFAULT = vector(0, 4*SIZE_UNIT, -1.5*SIZE_UNIT)
CAMERA_POS_DEFAULT = vector(0.5*SIZE_UNIT, -3*SIZE_UNIT, 2*SIZE_UNIT)
```

# Helper

```python
def setup_obj():
    ball = sphere(radius =BALL_RADIUS, 
                  color = BALL_COLOR, 
                  opacity = BALL_OPACITY)
    dir_vector_x = arrow(axis=VECTOR_U_X,
                         length = DIR_VECTOR_LENGTH,
                         color = DIR_VECTOR_X_COLOR)
    dir_vector_y = arrow(axis=VECTOR_U_Y, 
                         length = DIR_VECTOR_LENGTH,
                         color = DIR_VECTOR_Y_COLOR)
    dir_vector_z = arrow(axis=VECTOR_U_Z, 
                         length = DIR_VECTOR_LENGTH,
                         color = DIR_VECTOR_Z_COLOR)    
    obj = compound([ball, dir_vector_x, dir_vector_y, dir_vector_z],
                     origin = VECTOR_O,  pos = VECTOR_O,
                     axis = BALL_AXIS, up=BALL_UP)
    return obj
```

# Visualization Object

```python
class VizObject:
    def __init__(self):
        scene.background = color.gray(0.5)  
        self._obj = setup_obj()
        self._axis = VECTOR_U_Z

    def show(self):
        return self._obj  

    def reset_camera(self):
        scene.camera.pos = CAMERA_POS_DEFAULT
        scene.camera.axis = CAMERA_AXIS_DEFAULT
        
    def set_orientation(self, aPhi, aTheta, aPsi): 
        cphi, sphi = cos(aPhi), sin(aPhi)
        ctheta, stheta = cos(aTheta), sin(aTheta)
        cpsi, spsi = cos(aPsi), sin(aPsi)        
        R = np.array([
            [ctheta*cpsi, ctheta*spsi, -stheta],
            [sphi*stheta*cpsi-cphi*spsi, sphi*stheta*spsi+cphi*cpsi, sphi*ctheta],
            [cphi*stheta*cpsi+sphi*spsi, cphi*stheta*spsi-sphi*cpsi, cphi*ctheta]
        ])
        
        new_axis = R.dot([BALL_AXIS.x, BALL_AXIS.y, BALL_AXIS.z])
        new_up = R.dot([BALL_UP.x, BALL_UP.y, BALL_UP.z])
        
        self._obj.axis = vector(new_axis[0], new_axis[1], new_axis[2])
        self._obj.up = vector(new_up[0], new_up[1], new_up[2])
```

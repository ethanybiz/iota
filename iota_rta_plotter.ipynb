```python
PLOTTER_WINDOW_WIDTH_DP_MPU_DEFAULT = 300   

class MPURawDataPlotter(RealtimeSubWorker):
    DATA_CHANNELS = ['AccX', 'AccY', 'AccZ', 
                     'GyrX', 'GyrY', 'GyrZ']
    
    def __init__(self):
        super().__init__(aEndPtSrc=SOCK_TCP_MCUSRV_SUB,                           
                         aTopic='MPU')

        self._data_itp = PlotterMultiChannel(aChannelList = self.DATA_CHANNELS, 
                                             aWinWidth=PLOTTER_WINDOW_WIDTH_DP_MPU_DEFAULT)       

    
    def _on_message(self, aMsg):            
        self._data_itp.ingest_str(aMsg)

    
    def _close(self):  
        del self._data_itp   
```

```python
rt_plotter = MPURawDataPlotter()
```

```python
rt_plotter.start_worker() 
```

# Connect to DB

```python
con = sqlite3.connect(DB_PATH_DEVICE_SQLITE) 
cursor = con.cursor()
```

# DDL Query

## Create Table

```python
CURR_QUERY = f'''
CREATE TABLE IF NOT EXISTS {DB_TBL_NAME_RAW_DATA} (
    tp_srv_in TIMESTAMP not null
    , id_srv INTEGER not null        
    , id_unit varchar(255) not null
    , id_pkg INTEGER not null
    , tick_unit INTEGER not NULL
    , data_type varchar(16) not null
    , data_line varchar(1023) not null
);
'''

query_result = cursor.execute(CURR_QUERY)
con.commit()
print(query_result)
```

## Insert then Delete recoard

```python
CURR_QUERY = f'''
INSERT INTO {DB_TBL_NAME_RAW_DATA} VALUES (
'2025-04-01 12:12:12', 1212, 'TestDeviceName', 12, 121212, 'DM',
'{{"AccX":-3.36, "AccY":-6.19, "AccZ":5.18, "GyrX":-0.05, "GyrY":0.06, "GyrY":-0.02, "TmpC":19.10}}'
)
'''

query_result = cursor.execute(CURR_QUERY)
con.commit()
print(query_result)
```

```python
CURR_QUERY = f'''
DELETE FROM {DB_TBL_NAME_RAW_DATA} 
WHERE id_srv = 1212
'''

query_result = cursor.execute(CURR_QUERY)
con.commit()
print(query_result)
```

# DML Querys

```python
CURR_QUERY = f'''
SELECT COUNT(*)
FROM {DB_TBL_NAME_RAW_DATA}
;
'''

query_result = pd.read_sql_query(CURR_QUERY, con)
print(query_result.shape[0])
display(query_result)
```

```python
CURR_QUERY = f'''
SELECT *
FROM {DB_TBL_NAME_RAW_DATA}
;
'''

query_result = pd.read_sql_query(CURR_QUERY, con)
print(query_result.shape[0])
display(query_result)
```

# Close DB

```python
class RotationSimulator(RealtimeSubWorker):
    def __init__(self):     
        super().__init__(aEndPtSrc=SOCK_TCP_MCUSRV_SUB, 
                         aTopic='MPU')

        self._data_tfm_calibrator = CalibrateWithSphereAlg()
        self._data_tfm_rotator = RotationTransformerMahony()
               
        self._data_itp = VizObject()
        self._data_itp.show()
        self._data_itp.reset_camera()

        
    def _on_message(self, aMsg):
        aMsg = json.loads(aMsg)
        aMsg = {**{'tick':aMsg['tick']}, **aMsg['metrics']}        
        tmp = self._data_tfm_calibrator.apply(aMsg)        
        tmp = self._data_tfm_rotator.apply(tmp)            
        self._data_itp.set_orientation(aPhi = tmp['phi'], 
                                       aTheta= tmp['theta'],
                                       aPsi=tmp['psi'])    
        

    def _close(self):  
         del self._data_itp  
```

# Run Realtime 3D Simulatorm

```python
rt_pose_simulator = RotationSimulator()
```

```python
rt_pose_simulator.start_worker() 
```
